```python
# Dictonary 
#empty 
dic = {}

print(type(dic))

dic1 = {'name': 'kamal', 'age':20, 'address': 'Dhaka', 'number':[3,4,5,6,7]}
print(type(dic1), dic1)
```

```python
## Accessing value of dictonary 

print(dic1['name'])
print(dic1.get('age'))

# update element
dic1['name'] = 'jamal'

print(dic1)

# new value add in dictonary 
dic1['math_marks'] = 30

print(dic1)
```

```python
# deleting items form dictonary 
del dic1['math_marks']
 
print(dic1)
```
