```python
input_string = input().strip().lower();
words = " ".join(input_string).split()

sorted_words =  sorted(words)

# print(sorted_words)

count = {}

for word in sorted_words:
    count[word] = count.get(word, 0) + 1

# print(sorted(count))
# print(count)
# print(words)
for key , value in count.items():
    print(f"{key} : {value}")
```

```python
from collections import Counter
import sys

data = sys.stdin.read().lower()  # reads full input once
count = Counter(data)            # counts each character

for char in sorted(count):
    print(f"{char} : {count[char]}")
```

```python
input_string = input().strip().lower()
count = {}
for char in input_string:
    count[char] = count.get(char, 0) + 1

for char in sorted(count.keys()):
    print(f"{char} : {count[char]}")
```

```python
# sets 

input_string = input().strip()
n, m = map(int, input_string.split())
n_array = [int(x) for x in input().split()]

a_set =  {int(x ) for x in input().split()}
b_set = {int(x ) for x in input().split()}

happiness = 0;


for i in  n_array:

    if i in a_set:
        happiness += 1
    elif i  in b_set:
        happiness -= 1


# print(n, m)
# print(n_array)

# print(a_set)
# print(b_set)

print(happiness)
```
