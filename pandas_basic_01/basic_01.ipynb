## Pandas Series তৈরি করা (Creating a Series)

### **Pandas এর প্রথম কোর ডেটা স্ট্রাকচার - Series তৈরি করা।**

### ***Series হলো একটি এক-মাত্রিক (one-dimensional) লেবেলযুক্ত অ্যারে, যা অনেকটা ডিকশনারি বা একটি এক্সেল কলামের মতো।***


```python
import pandas as pd
# a list of integer data to create a Series
data = [10, 20, 30, 40, 50]

my_series = pd.Series(data)
print(my_series)

## dictonary data থেকে Series তৈরি করা
data_dict = {'a': 100, 'b': 200, 'c': 300}
my_series_from_dict = pd.Series(data_dict)
print(my_series_from_dict)
```

## Pandas এর সবচেয়ে গুরুত্বপূর্ণ ডেটা স্ট্রাকচার - DataFrame তৈরি করা।
### ***DataFrame হলো একটি দ্বি-মাত্রিক (two-dimensional) লেবেলযুক্ত ডেটা স্ট্রাকচার, যা সারি (rows) এবং কলাম (columns) দ্বারা গঠিত। এটি একটি সম্পূর্ণ এক্সেল শিট বা SQL টেবিলের মতো।***

```python
import pandas as pd
## dictionary data থেকে DataFrame তৈরি করা
student_data = {
    'Subject': ['math', 'english', 'bangla'],
    'Mark': [80, 90, 80],
    'Name': ['kamal', 'jamal', 'tomal']
}
## create DataFrame
df = pd.DataFrame(student_data)
print(df)
## display DataFrame info
print(df.info())
print(df.describe())
print(df.head())  # প্রথম ৫টি রেকর্ড দেখাবে
print(df.tail())  # শেষের ৫টি রেকর্ড দেখাবে 
print(df.columns)  # কলাম নাম দেখাবে
print(df.index)    # index দেখাবে  
print(df.dtypes)   # প্রতিটি কলামের ডেটা টাইপ দেখাবে
print(df.shape)    # DataFrame এর আকার (rows, columns) দেখাবে
print(df.values)   # DataFrame এর ডেটা অ্যারে আকারে দেখাবে
print(df['Name'])  # নির্দিষ্ট একটি কলামের ডেটা দেখাবে
print(df[['Name', 'Mark']])  # একাধিক কলামের ডেটা দেখাবে
print(df.iloc[0])  # প্রথম সারির ডেটা দেখাবে
print(df.iloc[:, 1])  # দ্বিতীয় কলামের ডেটা দেখাবে
print(df.loc[0, 'Name'])  # প্রথম সারির 'Name' কলামের ডেটা দেখাবে
print(df[df['Mark'] > 80])  # 'Mark' কলামে 80 এর বেশি মানের সারি দেখাবে
print(df.sort_values(by='Mark', ascending=False))  # 'Mark' কলাম অনুযায়ী ডেটা সাজাবে
print(df.groupby('Mark').size())  # 'Mark' কলাম অনুযায়ী গ্রুপিং করে সাইজ দেখাবে
print(df.isnull().sum())  # প্রতিটি কলামে null মানের সংখ্যা দেখাবে
print(df.drop_duplicates())  # ডুপ্লিকেট রেকর্ড বাদ দিয়ে ডেটা দেখাবে
print(df.reset_index(drop=True))  # index রিসেট করবে
print(df.set_index('Name'))  # 'Name' কলামকে index হিসেবে সেট করবে 
print(df.to_csv('students.csv', index=False))  # DataFrame কে CSV ফাইলে সেভ করবে
#print(df.to_excel('students.xlsx', index=False))  # DataFrame কে Excel ফাইলে সেভ করবে
print(pd.read_csv('students.csv'))  # CSV ফাইল থেকে DataFrame লোড করবে
#print(pd.read_excel('students.xlsx'))  # Excel ফাইল থেকে DataFrame লোড করবে
```

## DataFrame হলো Pandas এর মূল কাঠামো এবং এটি ডিকশনারি থেকে খুব সহজে তৈরি করা যায়।

## **বাস্তব জগতের ডেটাসেট (যেমন CSV ফাইল) থেকে ডেটা পড়া এবং DataFrame কে ফাইলে সেভ করা।**
***মেশিন লার্নিং-এ আমাদের প্রায়ই CSV (Comma-Separated Values) ফাইল থেকে ডেটা লোড করতে হয়।***

## pd.read_csv() এবং df.to_csv() ফাংশন ব্যবহার করে ফাইল থেকে ডেটা লোড এবং সেভ করা যায়।

```python
import pandas as pd

# একটি CSV ফাইল থেকে ডেটা পড়া
# ধরুন আপনার কাছে 'data.csv' নামে একটি ফাইল আছে
# df = pd.read_csv('data.csv') 

# আপাতত আমরা একটি DataFrame তৈরি করে সেভ করবো
data = {'City': ['Dhaka', 'Chittagong', 'Khulna'], 'Population': [21, 10, 2]}
df_city = pd.DataFrame(data)

# DataFrame কে CSV ফাইলে লেখা (সেভ করা)
# index=False দিলে ইনডেক্স কলামটি সেভ হবে না
df_city.to_csv('cities.csv', index=False)

# সেভ করা ফাইলটি আবার পড়া
df_read = pd.read_csv('cities.csv')
print("ফাইল থেকে পড়া ডেটা:")
print(df_read)
```


## DataFrame থেকে নির্দিষ্ট কলাম, সারি বা ডেটা পয়েন্ট সিলেক্ট করা।

**ডেটা অ্যানালাইসিসের জন্য নির্দিষ্ট ডেটা অ্যাক্সেস করা খুব জরুরি। এর জন্য .loc (লেবেল ভিত্তিক) এবং .iloc (ইনডেক্স ভিত্তিক) ব্যবহার করা হয়।**
