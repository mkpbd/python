## Pandas এর সবচেয়ে গুরুত্বপূর্ণ ডেটা স্ট্রাকচার - DataFrame তৈরি করা।
### ***DataFrame হলো একটি দ্বি-মাত্রিক (two-dimensional) লেবেলযুক্ত ডেটা স্ট্রাকচার, যা সারি (rows) এবং কলাম (columns) দ্বারা গঠিত। এটি একটি সম্পূর্ণ এক্সেল শিট বা SQL টেবিলের মতো।***

````python
import pandas as pd
## dictionary data থেকে DataFrame তৈরি করা
student_data = {
    'Subject': ['math', 'english', 'bangla'],
    'Mark': [80, 90, 80],
    'Name': ['kamal', 'jamal', 'tomal']
}
## create DataFrame
df = pd.DataFrame(student_data)
print(df)
## display DataFrame info
print(df.info())
print(df.describe())
print(df.head())  # প্রথম ৫টি রেকর্ড দেখাবে
print(df.tail())  # শেষের ৫টি রেকর্ড দেখাবে 
print(df.columns)  # কলাম নাম দেখাবে
print(df.index)    # index দেখাবে  
print(df.dtypes)   # প্রতিটি কলামের ডেটা টাইপ দেখাবে
print(df.shape)    # DataFrame এর আকার (rows, columns) দেখাবে
print(df.values)   # DataFrame এর ডেটা অ্যারে আকারে দেখাবে
print(df['Name'])  # নির্দিষ্ট একটি কলামের ডেটা দেখাবে
print(df[['Name', 'Mark']])  # একাধিক কলামের ডেটা দেখাবে
print(df.iloc[0])  # প্রথম সারির ডেটা দেখাবে
print(df.iloc[:, 1])  # দ্বিতীয় কলামের ডেটা দেখাবে
print(df.loc[0, 'Name'])  # প্রথম সারির 'Name' কলামের ডেটা দেখাবে
print(df[df['Mark'] > 80])  # 'Mark' কলামে 80 এর বেশি মানের সারি দেখাবে
print(df.sort_values(by='Mark', ascending=False))  # 'Mark' কলাম অনুযায়ী ডেটা সাজাবে
print(df.groupby('Mark').size())  # 'Mark' কলাম অনুযায়ী গ্রুপিং করে সাইজ দেখাবে
print(df.isnull().sum())  # প্রতিটি কলামে null মানের সংখ্যা দেখাবে
print(df.drop_duplicates())  # ডুপ্লিকেট রেকর্ড বাদ দিয়ে ডেটা দেখাবে
print(df.reset_index(drop=True))  # index রিসেট করবে
print(df.set_index('Name'))  # 'Name' কলামকে index হিসেবে সেট করবে 
print(df.to_csv('students.csv', index=False))  # DataFrame কে CSV ফাইলে সেভ করবে
#print(df.to_excel('students.xlsx', index=False))  # DataFrame কে Excel ফাইলে সেভ করবে
print(pd.read_csv('students.csv'))  # CSV ফাইল থেকে DataFrame লোড করবে
#print(pd.read_excel('students.xlsx'))  # Excel ফাইল থেকে DataFrame লোড করবে
````

## শর্তসাপেক্ষে ডেটা ফিল্টারিং (Conditional Filtering)
**নির্দিষ্ট শর্তের উপর ভিত্তি করে ডেটা ফিল্টার করা।**
#### *ব্র্যাকেটের মধ্যে শর্ত (df['column'] > value) দিয়ে খুব সহজে ডেটা ফিল্টার করা যায়। একাধিক শর্তের জন্য & (and) এবং | (or)  bitwise not ~  ব্যবহার হয়।*

````python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],   
    'Age': [28, 34, 29, 42],
    'Salary': [50000, 60000, 55000, 70000],
    'Department': ['HR', 'Finance', 'IT', 'Marketing']
}
df = pd.DataFrame(data)
# শর্তসাপেক্ষে ডেটা ফিল্টারিং
filtered_df = df[df['Age']>30]
print("Filtered DataFrame (Age > 30):")
print(filtered_df)
high_salary_df = df[df['Salary'] >= 60000]
print("Filtered DataFrame (Salary >= 60000):")  
print(high_salary_df)
marketing_dept_df = df[df['Department'] == 'Marketing']
print("Filtered DataFrame (Department == 'Marketing'):")    
print(marketing_dept_df)

## শর্তসাপেক্ষে একাধিক শর্ত ব্যবহার করে ডেটা ফিল্টারিং
age_salary_df = df[(df['Age'] > 30) & (df['Salary'] > 55000)]
print("Filtered DataFrame (Age > 30 and Salary > 55000):")
print(age_salary_df)
````

## অনুপস্থিত ডেটা হ্যান্ডলিং (Handling Missing Data)

ডেটাসেটে NaN (Not a Number) বা খালি ভ্যালু খুঁজে বের করা এবং তা ম্যানেজ করা।
বাস্তব ডেটাসেটে প্রায়ই কিছু ভ্যালু অনুপস্থিত থাকে। এগুলোকে হয় ফেলে দিতে হয় (dropna) অথবা কোনো মান দিয়ে পূরণ করতে হয় (fillna)।
***isnull().sum() দিয়ে অনুপস্থিত ডেটা খুঁজে dropna() দিয়ে বাদ দেওয়া বা fillna() দিয়ে পূরণ করা যায়।***



````python
import pandas as pd
import numpy as np
data ={
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': [9, 10, 11, 12],
    'D': [np.nan, 14, 15, 16]
}
df = pd.DataFrame(data)
print("মূল DataFrame:")
print(df)
# অনুপস্থিত ডেটা চেক করা
print("প্রতিটি কলামে অনুপস্থিত ডেটার সংখ্যা:")
print(df.isnull().sum())
# অনুপস্থিত ডেটা ফিল্টার করা
# অনুপস্থিত ডেটাযুক্ত সারি বাদ দেওয়া
filtered_df = df.dropna()
print("অনুপস্থিত ডেটা ফিল্টার করার পর DataFrame:")
print(filtered_df)
# অনুপস্থিত ডেটা নির্দিষ্ট মান দিয়ে পূরণ করা (যেমন, গড় মান দিয়ে)
filled_df = df.fillna(df.mean())
print("অনুপস্থিত ডেটা পূরণ করার পর DataFrame:")
print(filled_df)
````

## ডেটা অপারেশন এবং নতুন কলাম তৈরি (Operations & Creating New Columns)
### বিদ্যমান কলামের উপর ভিত্তি করে নতুন কলাম তৈরি করা।
#### **কলামগুলোর মধ্যে গাণিতিক অপারেশন চালানো যায় বা ফাংশন প্রয়োগ করে নতুন তথ্য তৈরি করা যায়।**
 সরাসরি গাণিতিক অপারেশন বা .apply() ফাংশন ব্যবহার করে DataFrame এ নতুন কলাম যোগ করা যায়।


````python
import pandas as pd
df = pd.DataFrame({
    'পণ্য': ['আম', 'জাম', 'লিচু'],
    'দাম (কেজি)': [100, 120, 250],
    'পরিমাণ (কেজি)': [5, 3, 2]
})
print("মূল DataFrame:")
# print(df)
# নতুন কলাম যোগ করা
df['মোট মূল্য'] = df['দাম (কেজি)'] * df['পরিমাণ (কেজি)']
print("নতুন কলাম যোগ করার পর DataFrame:")
print(df)
# df['বর্ণনা'] = ['মিষ্টি আম', 'রসালো জাম', 'মধুর লিচু']

# apply এবং lambda ফাংশন ব্যবহার করে নতুন কলাম
# ধরুন, দামের উপর ১০% ভ্যাট যোগ হবে
df['ভ্যাটসহ দাম'] = df['মোট মূল্য'].apply(lambda x: x * 1.10)
print("\nভ্যাটসহ দামের কলাম:\n", df)
````

## গ্রুপিং এবং অ্যাগ্রিগেশন (Grouping and Aggregation)

### **ডেটাকে গ্রুপে ভাগ করে প্রতি গ্রুপের উপর পরিসংখ্যান চালানো। এটি GROUP BY নামে পরিচিত।**
এটি ডেটা অ্যানালাইসিসের অন্যতম শক্তিশালী একটি ফিচার। যেমন - "প্রতিটি ডিপার্টমেন্টে গড় বেতন কত?

#### groupby() ব্যবহার করে ডেটাকে বিভিন্ন গ্রুপে ভাগ করে .mean(), .sum(), .count() এর মতো অ্যাগ্রিগেট ফাংশন চালানো যায়।


````python
import pandas as pd
data ={
    'Department': ['HR', 'Finance', 'IT', 'HR', 'Finance', 'IT'],
    'Salary': [50000, 60000, 55000, 52000, 62000, 58000],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank']
}
df_employees = pd.DataFrame(data)
## ডিপার্টমেন্ট অনুযায়ী গ্রুপ করা এবং গড় বেতন বের করা
agv_salaray_by_dept = df_employees.groupby('Department')['Salary'].mean()
print(agv_salaray_by_dept)

## প্রতিটি ডিপার্টমেন্টে সর্বোচ্চ বেতন কার?
max_salary_by_dept = df_employees.groupby('Department')['Salary'].max()
print(max_salary_by_dept)
## প্রতিটি ডিপার্টমেন্টে কর্মচারীর সংখ্যা কত?
employee_count_by_dept = df_employees.groupby('Department').size()
print(employee_count_by_dept)
# প্রতিটি ডিপার্টমেন্টে মোট বেতন কত?
total_salary_by_dept = df_employees.groupby('Department')['Salary'].sum()
print(total_salary_by_dept)
# গ্রুপিং এবং একাধিক অ্যাগ্রিগেশন ফাংশন প্রয়োগ করা
agg_functions = df_employees.groupby('Department')['Salary'].agg(['mean', 'max', 'min', 'sum', 'count'])
print(agg_functions)

````

## ডেটাফ্রেম একত্রিত করা (Merging, Joining, Concatenating)
#### একাধিক DataFrame কে একসাথে জোড়া লাগানো।
 > Concatenate: DataFrame গুলোকে একটির নিচে আরেকটি বা পাশাপাশি যোগ করে।
 > Merge: SQL Join এর মতো, একটি কমন কলামের উপর ভিত্তি করে দুটি DataFrame কে একত্রিত করে।

#### pd.concat() ডেটাফ্রেমকে স্ট্যাক করে আর pd.merge() ডেটাবেসের join এর মতো কাজ করে।


````python
import pandas as pd
df1 = pd.DataFrame({ 'A': [1, 2, 3], 'B': [4, 5, 6] })
df2 = pd.DataFrame({ 'A': [7, 8, 9], 'B': [10, 11, 12] })
# concate (vertically ) Concat (একটির নিচে আরেকটি)
df_concat = pd.concat([df1, df2], ignore_index=True)
print("Concatenated DataFrame (vertically):")
print(df_concat)
# concate (horizontally) Concat (একটির পাশে আরেকটি)
df_concat_horiz = pd.concat([df1, df2], axis=1)
print("Concatenated DataFrame (horizontally):")
print(df_concat_horiz)

# merge (SQL style join)
left_df = pd.DataFrame({ 'Key': ['K1', 'K2', 'K3'], 'Left_Val': [1, 2, 3] })
right_df = pd.DataFrame({ 'Key': ['K1', 'K2', 'K4'], 'Right_Val': [4, 5, 6] })

# Mearge examples 
mearge_df = pd.merge(left_df, right_df, on='Key')
print("Merged DataFrame (default inner join):")
print(mearge_df)

# inner join
df_merged_inner = pd.merge(left_df, right_df, on='Key', how='inner')
print("Merged DataFrame (inner join):") 
print(df_merged_inner)
# outer join
df_merged_outer = pd.merge(left_df, right_df, on='Key', how='outer')
print("Merged DataFrame (outer join):")
print(df_merged_outer)
# left join
df_merged_left = pd.merge(left_df, right_df, on='Key', how='left')

print("Merged DataFrame (left join):")
print(df_merged_left)
# right join
df_merged_right = pd.merge(left_df, right_df, on='Key', how='right')
print("Merged DataFrame (right join):")
print(df_merged_right)

# merge with different column names
left_df2 = pd.DataFrame({ 'Left_Key': ['K1', 'K2', 'K3'], 'Left_Val': [1, 2, 3] })
right_df2 = pd.DataFrame({ 'Right_Key': ['K1', 'K2', 'K4'], 'Right_Val': [4, 5, 6] })
df_merged_diff_cols = pd.merge(left_df2, right_df2, left_on='Left_Key', right_on='Right_Key', how='inner')
print("Merged DataFrame with different column names:")
print(df_merged_diff_cols)
````

##  টাইম সিরিজ ডেটা নিয়ে কাজ (Working with Time Series)

 তারিখ এবং সময় সম্পর্কিত ডেটা নিয়ে কাজ করা।

অনেক ডেটাসেটেই সময় একটি গুরুত্বপূর্ণ ফিচার (যেমন - শেয়ার বাজার, আবহাওয়া)। Pandas টাইম সিরিজ ডেটা নিয়ে কাজ করার জন্য অসাধারণ।

pd.to_datetime() দিয়ে স্ট্রিংকে তারিখ ফরম্যাটে রূপান্তর করা যায় এবং সময়ভিত্তিক ডেটা খুব সহজে ফিল্টার করা যায়।

````python
import pandas as pd
# date time range তৈরি করা
date_rng = pd.to_datetime( pd.date_range('20250101', periods=5))
# date_rng প্রিন্ট করা
# print("Date Range:")
# print(date_rng)

df_time = pd.DataFrame({'date':date_rng, 'value':[10,200,3,4,500]})

print("Date Time DataFrame:")
print(df_time)
# date কলামকে index হিসেবে সেট করা
df_time.set_index('date', inplace=True)
print("Date Time DataFrame with date as index:")
print(df_time)

# নির্দিষ্ট তারিখের ডেটা সিলেক্ট করা
specific_date = df_time.loc['2025-01-03']
print("Data for specific date (2025-01-03):")
print(specific_date)
````

## ডেটা ক্লিনিং (Advanced Data Cleaning)
## ডুপ্লিকেট ডেটা, ডেটা টাইপ পরিবর্তন এবং স্ট্রিং ম্যানিপুলেশন।
#### বাস্তব ডেটা প্রায়ই অপরিষ্কার থাকে। ডুপ্লিকেট সারি বাদ দেওয়া, ভুল ডেটা টাইপ ঠিক করা এবং টেক্সট ডেটা পরিষ্কার করা জরুরি।

##### .drop_duplicates(), .str accessor এবং .astype() ব্যবহার করে ডেটাকে আরও ব্যবহারযোগ্য করে তোলা যায়।

### drop_duplicates(), .str accessor এবং .astype() ব্যবহার করে ডেটাকে আরও ব্যবহারযোগ্য করে তোলা যায়।

````python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Alice'],
    'Age': [28, 34, 30, 42, 30, 28],
    'Salary': [50000, 60000, 55000, 70000, 65000, 50000],
    'Department': ['HR', 'Finance', 'IT', 'Marketing', 'Finance', 'HR']
}
df = pd.DataFrame(data)
# 1 duplicate rows বাদ দেওয়া
df_no_duplicates = df.drop_duplicates()
print("DataFrame after dropping duplicate rows:")
print(df_no_duplicates)
# ২ নির্দিষ্ট কলাম অনুযায়ী duplicate rows বাদ দেওয়া
df_no_dup_name = df.drop_duplicates(subset=['Name'])
print("DataFrame after dropping duplicates based on 'Name' column:")
print(df_no_dup_name)
# ৩ ডেটা টাইপ পরিবর্তন করা
df['Age'] = df['Age'].astype(float)
print("DataFrame after changing 'Age' column to float:")
print(df.dtypes)
# ৪ স্ট্রিং ম্যানিপুলেশন
df['Name'] = df['Name'].str.upper()
print("DataFrame after converting 'Name' column to uppercase:")
print(df)

print(df_no_duplicates.info())
````

## পিভট টেবিল তৈরি করা (Pivot Tables)
groupby এর একটি অ্যাডভান্সড এবং ভিজ্যুয়াল রূপ - পিভট টেবিল।
পিভট টেবিল ডেটাকে একটি নতুন আঙ্গিকে সাজিয়ে সামারি তৈরি করে, যা এক্সেল ব্যবহারকারীদের কাছে খুব পরিচিত।

````python
import pandas as pd
data = {'Date': ['2023-01-01', '2023-01-01', '2023-01-02', '2023-01-02'],
        'City': ['Dhaka', 'Chittagong', 'Dhaka', 'Chittagong'],
        'Temperature': [22, 24, 23, 25],
        'Humidity': [60, 65, 58, 68]}
df_weather = pd.DataFrame(data)
# pivot table তৈরি করা
pivot_table = df_weather.pivot_table(values='Temperature',
                                     index='Date',
                                     columns='City',
                                     aggfunc='mean')
print("Pivot Table:")
print(pivot_table)
````

*   **Attribute:** একটি প্রপার্টি বা বৈশিষ্ট্য যা অবজেক্ট সম্পর্কে তথ্য দেয় এবং এটি কল করার জন্য `()` বন্ধনী লাগে না (যেমন: `df.shape`)।

#### **৪. ডেটা ক্লিনিং এবং ম্যানিপুলেশন (Data Cleaning & Manipulation)**

*   `df.dropna()`: অনুপস্থিত ডেটা (NaN) আছে এমন যেকোনো সারি বা কলাম বাদ দেয়।
*   `df.fillna(value)`: অনুপস্থিত ডেটা (NaN) একটি নির্দিষ্ট `value` দিয়ে পূরণ করে।
*   `df.astype(dtype)`: একটি কলামের ডেটা টাইপ পরিবর্তন করে (যেমন: `df['age'].astype(int)`)।
*   `df.rename(columns={'old_name': 'new_name'})`: কলামের নাম পরিবর্তন করে।
*   `df.drop('column_name', axis=1)`: নির্দিষ্ট কলাম (`axis=1`) বা সারি (`axis=0`) বাদ দেয়।

#### **১১. অ্যাডভান্সড ডেটা রিশেপিং এবং স্ট্রাকচারিং (Advanced Data Reshaping)**
এই মেথডগুলো ডেটাকে "wide" থেকে "long" ফরম্যাটে বা এর বিপরীতে রূপান্তর করতে ব্যবহৃত হয়, যা ডেটা অ্যানালাইসিস এবং ভিজ্যুয়ালাইজেশনের জন্য খুবই গুরুত্বপূর্ণ।

*   `df.pivot(index, columns, values)`: `pivot_table` এর মতোই, কিন্তু এটি ডেটা অ্যাগ্রিগেট করে না। প্রতিটি `index`/`column` জোড়ার জন্য একটি ইউনিক ভ্যালু থাকতে হয়।
*   `df.melt(id_vars, value_vars)`: `pivot` এর ঠিক বিপরীত কাজ করে। এটি একটি "wide" DataFrame-কে "long" ফরম্যাটে রূপান্তর করে। একাধিক কলামকে একটি "variable" কলাম এবং একটি "value" কলামে পরিণত করে।
    ```python
    df_wide = pd.DataFrame({'Subject': ['Math', 'Science'], 'Alice': [90, 88], 'Bob': [85, 92]})
    # df_wide looks "wide"
    df_long = df_wide.melt(id_vars=['Subject'], var_name='Student', value_name='Score')
    # df_long is now "long" and easier to group by student
    ```
*   `df.stack()`: DataFrame-এর কলামগুলোকে সারিতে (index-এ) নিয়ে আসে। এটি MultiIndex (হায়ারার্কিকাল ইনডেক্স) তৈরি করে।
*   `df.unstack()`: `stack` এর বিপরীত। এটি ইনডেক্সের একটি লেভেলকে কলামে রূপান্তর করে।
*   `df.explode('column_name')`: একটি কলামের মধ্যে থাকা লিস্ট-এর মতো আইটেমগুলোকে আলাদা আলাদা সারিতে ভাগ করে দেয়।
    ```python
    df_tags = pd.DataFrame({'PostID': [1, 2], 'Tags': [['python', 'pandas'], ['python', 'numpy']]})
    df_exploded = df_tags.explode('Tags')
    ```
*   `pd.crosstab(index, columns)`: দুটি ফ্যাক্টর বা কলামের মধ্যে ফ্রিকোয়েন্সি টেবিল তৈরি করে। এটি মূলত `pivot_table` এর একটি বিশেষ রূপ যা ফ্রিকোয়েন্সি গণনা করে।

#### **১২. উইন্ডো ফাংশন (Window Functions)**
এই ফাংশনগুলো ডেটার একটি নির্দিষ্ট "উইন্ডো" বা সাবসেটের উপর ক্যালকুলেশন চালায়। টাইম সিরিজ ডেটার জন্য এগুলো খুবই শক্তিশালী।

*   `df.rolling(window_size)`: একটি স্লাইডিং বা চলমান উইন্ডো তৈরি করে। এর উপর অ্যাগ্রিগেট ফাংশন চালানো হয়। যেমন, ৭ দিনের মুভিং এভারেজ বের করার জন্য।
    ```python
    df['Sales_7_Day_MA'] = df['Sales'].rolling(window=7).mean()
    ```
*   `df.expanding()`: একটি ক্রমবর্ধমান উইন্ডো তৈরি করে। এটি ডেটাসেটের শুরু থেকে বর্তমান পয়েন্ট পর্যন্ত সমস্ত ডেটা নিয়ে ক্যালকুলেশন করে। যেমন, কিউমুলেটিভ বা ক্রমযোজিত যোগফল বের করা।
    ```python
    df['Cumulative_Sales'] = df['Sales'].expanding().sum()
    ```
*   `df.ewm(...)`: এক্সপোনেনশিয়াল ওয়েটেড মুভিং (Exponentially Weighted Moving) ফাংশন। এটি সাম্প্রতিক ডেটা পয়েন্টগুলোকে বেশি গুরুত্ব দেয়। এটি সাধারণ মুভিং এভারেজের চেয়ে বেশি রেসপন্সিভ।
    ```python
    df['Sales_EWM'] = df['Sales'].ewm(span=7).mean()
    ```

    *   `.cat.codes`: প্রতিটি ক্যাটাগরির জন্য একটি নিউমেরিক্যাল কোড দেখায়।

    *   `.cat.reorder_categories([...])`: ক্যাটাগরিগুলোর ক্রম পরিবর্তন করে (প্লটিং এর জন্য উপকারী)।
    *   `.cat.rename_categories([...])`: ক্যাটাগরির নাম পরিবর্তন করে।

*   `pd.MultiIndex`: হায়ারার্কিকাল বা বহু-স্তরীয় ইনডেক্স তৈরি করার জন্য ব্যবহৃত হয়। `groupby` বা `pivot` করার সময় এটি স্বয়ংক্রিয়ভাবে তৈরি হতে পারে।
*   `df.xs(key, level)`: MultiIndex যুক্ত DataFrame থেকে নির্দিষ্ট লেভেলের ডেটা ক্রস-সেকশন বা স্লাইস করার জন্য একটি শক্তিশালী মেথড।
*   `df.query('expression')`: একটি স্ট্রিং এক্সপ্রেশন ব্যবহার করে DataFrame ফিল্টার করার একটি খুব সহজ এবং পাঠযোগ্য উপায়।

    df.query('age > 30 and city == "Dhaka"')

*   `df.mask(condition, value)`: যেখানে `condition` সত্য (True), সেখানকার ভ্যালুগুলোকে `value` দিয়ে প্রতিস্থাপন করে।
*   `df.where(condition, value)`: যেখানে `condition` মিথ্যা (False), সেখানকার ভ্যালুগুলোকে `value` দিয়ে প্রতিস্থাপন করে। এটি `mask`-এর বিপরীত।

#### **১৭. মেথড চেইনিং এবং `.pipe()` দিয়ে পরিচ্ছন্ন কোড লেখা (Method Chaining & Clean Code with `.pipe()`)**
মেথড চেইনিং হলো একটির পর একটি মেথডকে চেইন আকারে কল করা। এটি কোডকে সংক্ষিপ্ত করে, কিন্তু অনেক লম্বা হয়ে গেলে পড়া কঠিন হয়ে যায়। এখানেই `.pipe()` মেথডটি কাজে আসে।

**ব্যাখ্যা:**
`.pipe()` আপনাকে একটি ফাংশনকে পুরো DataFrame-এর উপর প্রয়োগ করতে দেয়, যা মেথড চেইনিং-এর মধ্যে ব্যবহার করা যায়। এটি কোডকে অত্যন্ত পাঠযোগ্য (readable) এবং মডুলার করে তোলে।

**কোড উদাহরণ:**
ধরুন, আমাদের কিছু ডেটা ক্লিনিং ধাপ আছে: অনুপস্থিত ভ্যালু পূরণ করা, বয়স ৩০-এর বেশি মানুষদের ফিল্টার করা, এবং একটি নতুন কলাম যোগ করা।

```python
import pandas as pd
import numpy as np

def fill_missing_age(df):
    """বয়সের অনুপস্থিত ভ্যালু গড় দিয়ে পূরণ করে"""
    df['Age'].fillna(df['Age'].mean(), inplace=True)
    return df

def filter_by_age(df, threshold):
    """নির্দিষ্ট বয়সের বেশি মানুষদের ফিল্টার করে"""
    return df[df['Age'] > threshold]

def add_age_category(df):
    """বয়সের উপর ভিত্তি করে ক্যাটাগরি যোগ করে"""
    df['Age_Group'] = pd.cut(df['Age'], bins=[0, 30, 50, 100], labels=['Young', 'Adult', 'Senior'])
    return df

# ডেটাসেট তৈরি
data = {'Name': ['A', 'B', 'C', 'D'], 'Age': [25, np.nan, 45, 55]}
df = pd.DataFrame(data)

# .pipe() ব্যবহার করে ক্লিনিং পাইপলাইন
cleaned_df = (df.pipe(fill_missing_age)
                .pipe(filter_by_age, threshold=30) # .pipe() ফাংশনে আর্গুমেন্ট পাস করা
                .pipe(add_age_category)
             )

print(cleaned_df)
```
**শিক্ষা:** `.pipe()` ব্যবহার করে ডেটা প্রসেসিংয়ের ধাপগুলোকে আলাদা আলাদা ফাংশনে ভাগ করে একটি পরিচ্ছন্ন এবং রক্ষণাবেক্ষণযোগ্য (maintainable) পাইপলাইন তৈরি করা যায়।

#### **১৮. পারফরম্যান্স: `.apply()` এর পরিবর্তে ভেক্টরাইজেশন (Performance: Vectorization over `.apply()`)**
বড় ডেটাসেটে, সারি বরাবর লুপ চালানো (যেমন `.apply(axis=1)`) খুবই ধীরগতির হতে পারে। এর পরিবর্তে, Pandas এবং NumPy-এর বিল্ট-ইন "ভেক্টরাইজড" অপারেশন ব্যবহার করা উচিত, যা পুরো কলাম বা Series-এর উপর একসাথে কাজ করে।

**ব্যাখ্যা:**
ভেক্টরাইজেশন হলো ডেটার সম্পূর্ণ অ্যারে বা সিরিজের উপর অপারেশন চালানো, যা底层 C বা Fortran কোডে এক্সিকিউট হওয়ায় অনেক দ্রুত হয়।

**কোড উদাহরণ:**
ধরুন, আমরা বেতনের উপর ভিত্তি করে ট্যাক্স গণনা করতে চাই।

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'Salary': np.random.randint(30000, 100000, size=100000)})

# পদ্ধতি ১: .apply() (ধীরগতির)
def calculate_tax(salary):
    if salary > 70000:
        return salary * 0.20
    else:
        return salary * 0.10

# %timeit df['Tax_Apply'] = df['Salary'].apply(calculate_tax) # এটি জুপিটারে চালালে সময় দেখাবে

# পদ্ধতি ২: ভেক্টরাইজেশন (অত্যন্ত দ্রুত)
# np.where(condition, value_if_true, value_if_false)
# %timeit df['Tax_Vectorized'] = np.where(df['Salary'] > 70000, df['Salary'] * 0.20, df['Salary'] * 0.10)

# এখানে শুধু আউটপুট দেখাচ্ছি
df['Tax_Vectorized'] = np.where(df['Salary'] > 70000, df['Salary'] * 0.20, df['Salary'] * 0.10)
print(df.head())
```
**শিক্ষা:** যখনই সম্ভব, লুপ বা `.apply()` এড়িয়ে চলুন এবং NumPy-এর কন্ডিশনাল ফাংশন (`np.where`) বা Pandas-এর বিল্ট-ইন গাণিতিক অপারেশন ব্যবহার করুন।

#### **১৯. জটিল ডেটা টাইপ নিয়ে কাজ: JSON এবং নেস্টেড ডেটা (Handling Complex Data Types: JSON & Nested Data)**
API থেকে পাওয়া ডেটা প্রায়ই JSON ফরম্যাটে এবং নেস্টেড (nested) বা স্তরযুক্ত থাকে। Pandas এগুলোকে সহজে ফ্ল্যাট DataFrame-এ পরিণত করতে পারে।

**ব্যাখ্যা:**
`pd.json_normalize()` ফাংশনটি নেস্টেড ডিকশনারি এবং লিস্টকে ভেঙে আলাদা আলাদা কলামে পরিণত করে।

**কোড উদাহরণ:**
```python
import pandas as pd

# একটি নেস্টেড JSON ডেটার উদাহরণ
data = [
    {'id': 1, 'student': {'name': 'Alice', 'major': 'Physics'}},
    {'id': 2, 'student': {'name': 'Bob', 'major': 'Chemistry'}},
]

# সরাসরি DataFrame তৈরি করলে student কলামে ডিকশনারি থাকবে
df_nested = pd.DataFrame(data)
print("নেস্টেড DataFrame:\n", df_nested)

# json_normalize ব্যবহার করে ফ্ল্যাট করা
df_flat = pd.json_normalize(data)
print("\nফ্ল্যাট করা DataFrame:\n", df_flat)
```
**শিক্ষা:** `pd.json_normalize()` ব্যবহার করে জটিল JSON ডেটাকে সহজেই একটি ব্যবহারযোগ্য টেবুলার ফরম্যাটে আনা যায়।

#### **২০. অ্যাডভান্সড টাইম সিরিজ কন্ট্রোল (Advanced Time Series Control)**
টাইম সিরিজ ডেটা নিয়ে কাজ করার সময় টাইমজোন (Timezone), ডিউরেশন (Duration) এবং কাস্টম ফ্রিকোয়েন্সি খুব গুরুত্বপূর্ণ।

*   `tz_localize()` / `tz_convert()`: একটি টাইম সিরিজকে টাইমজোন সম্পর্কে "সচেতন" (aware) করে এবং এক টাইমজোন থেকে অন্যটিতে রূপান্তর করে।
*   `pd.Timedelta`: দুটি সময়ের মধ্যে পার্থক্য বা একটি নির্দিষ্ট সময়কাল বোঝাতে ব্যবহৃত হয়।
*   `pd.offsets`: বিশেষায়িত সময় অফসেট তৈরি করে, যেমন মাসের শেষ দিন (`MonthEnd`), বা ব্যবসায়িক দিন (`BDay`)।

**কোড উদাহরণ:**
```python
import pandas as pd

s = pd.Series(pd.to_datetime(['2023-10-26 09:00:00', '2023-10-27 10:00:00']))

# 1. টাইমজোন যোগ করা এবং রূপান্তর করা
s_utc = s.dt.tz_localize('UTC')
s_dhaka = s_utc.dt.tz_convert('Asia/Dhaka')
print("ঢাকা টাইমজোনে:\n", s_dhaka)

# 2. Timedelta ব্যবহার করে সময় যোগ করা
future_time = s_dhaka + pd.Timedelta(days=1, hours=5)
print("\nভবিষ্যতের সময়:\n", future_time)

# 3. মাসের শেষ দিন বের করা
month_end_date = s_dhaka + pd.offsets.MonthEnd(0) # 0 মানে বর্তমান মাস
print("\nমাসের শেষ দিন:\n", month_end_date)
```
**শিক্ষা:** `.dt` accessor-এর সাথে এই অ্যাডভান্সড ফিচারগুলো ব্যবহার করে জটিল সময়-ভিত্তিক গণনা এবং ম্যানিপুলেশন করা যায়।

#### **২১. সাধারণ ভুল এবং ওয়ার্নিং বোঝা (Understanding Common Pitfalls & Warnings)**
Pandas ব্যবহার করার সময় কিছু সাধারণ ভুল এবং ওয়ার্নিং দেখা যায় যা নতুনদের বিভ্রান্ত করতে পারে।

*   **`SettingWithCopyWarning`**: এটি সম্ভবত সবচেয়ে সাধারণ ওয়ার্নিং। এটি তখন ঘটে যখন আপনি একটি DataFrame-এর স্লাইস বা কপির উপর মান নির্ধারণ করার চেষ্টা করেন।
    *   **ভুল পদ্ধতি (Chained Indexing):** `df[df['A'] > 0]['B'] = 100` # এখানে ওয়ার্নিং আসতে পারে
    *   **সঠিক পদ্ধতি (`.loc`):** `df.loc[df['A'] > 0, 'B'] = 100` # এটি সরাসরি মূল DataFrame-এ কাজ করে

*   **`object` ডেটা টাইপ**: যখন একটি কলামে একাধিক ধরনের ডেটা (যেমন সংখ্যা এবং স্ট্রিং) মিশে থাকে, তখন Pandas সেটিকে `object` টাইপ হিসেবে চিহ্নিত করে। `object` কলামে ভেক্টরাইজড অপারেশন চালানো যায় না এবং এটি অনেক বেশি মেমরি ব্যবহার করে। সবসময় `.info()` দিয়ে চেক করুন এবং `.astype()` দিয়ে সঠিক টাইপে রূপান্তর করার চেষ্টা করুন।
