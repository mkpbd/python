B. Even Array
সমস্যার মূল ধারণা
আমাদের এমন একটা অ্যারে দরকার, যেখানে
👉 ইনডেক্স (index) জোড় হলে, সেই ইনডেক্সে থাকা সংখ্যাটাও জোড় (even) হবে
👉 ইনডেক্স (index) বিজোড় হলে, সেই ইনডেক্সে থাকা সংখ্যাটাও বিজোড় (odd) হবে

অর্থাৎ parity (জোড়-বিজোড় অবস্থা) — ইনডেক্স আর ভ্যালুর — এক হতে হবে।
এটাই “good array”।

সারসংক্ষেপ অ্যালগরিদম

1️⃣ Count wrongEven = কয়টি even ইনডেক্সে odd সংখ্যা আছে
2️⃣ Count wrongOdd = কয়টি odd ইনডেক্সে even সংখ্যা আছে
3️⃣ যদি wrongEven == wrongOdd হয় → উত্তর = wrongEven
4️⃣ না হলে → উত্তর = -1

```python
input_number = int(input())

for _ in range(input_number):
    n = int(input())
    arr = list(map(int,input().split()))

    wrong_even = 0 # even index , odd value 
    wrong_odd = 0 # odd index , even value 

    for i in range(n):
        if(i % 2 != arr[i] % 2):
            if i% 2 == 0:
                wrong_even += 1
            else :
                wrong_odd += 1
    if wrong_even == wrong_odd:
        print(wrong_even)
    else :
        print(-1)

```
