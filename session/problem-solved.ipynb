```python
import math

# Base class
class Shape:
    def area(self):
        return 0

# Subclass for Rectangle
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h
    
    def area(self):
        return self.w * self.h

# Subclass for Circle
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    
    def area(self):
        return math.pi * (self.r ** 2)

# Read input
data = input().split()
shape_type = data[0]

# Determine which shape to create
if shape_type == "Rectangle":
    w, h = map(float, data[1:])
    shape = Rectangle(w, h)
elif shape_type == "Circle":
    r = float(data[1])
    shape = Circle(r)
else:
    shape = Shape()

# Print area rounded to 2 decimals
print(f"Area: {shape.area():.2f}")
```

```python
import math
def strong_number (number):
    copy_orginal_number = number
    fac_sum = 0;

    ## looping 

    while number > 0:
        # get last digit
        digit = number % 10 
        # add factorial of digit
        fac_sum += math.factorial(digit)
        number //= 10

    # check the strong number 
    if fac_sum == copy_orginal_number:
        print("Strong Number")
    else :
        print("Not Strong Number")

number = int(input().strip())

strong_number(number)
```
