# Strings
Strings in python are surrounded by either single quotation marks, or double quotation marks.

'hello' is the same as "hello".

You can display a string literal with the print() function:

```python
print("Hello")
print('Hello')
```

# Quotes Inside Quotes
You can use quotes inside a string, as long as they don't match the quotes surrounding the string:

```python
print("It's alright")
print("He is called 'Johnny'")
print('He is called "Johnny"')
```

# Assign String to a Variable
Assigning a string to a variable is done with the variable name followed by an equal sign and the string:

```python
a = "Hello"
print(a)
```

# Multiline Strings
You can assign a multiline string to a variable by using three quotes:

```python
a = """Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua."""
print(a)
```

```python
a = '''Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua.'''
print(a)
```

# Strings are Arrays
Like many other popular programming languages, strings in Python are arrays of unicode characters.

However, Python does not have a character data type, a single character is simply a string with a length of 1.

Square brackets can be used to access elements of the string.

```python
a = "Hello, World!"
print(a[1])
```

# Looping Through a String
Since strings are arrays, we can loop through the characters in a string, with a for loop.

```python
for x in "banana":
  print(x)
```

# String Length
To get the length of a string, use the len() function.

```python
a = "Hello, World!"
print(len(a))
```

# Check String
To check if a certain phrase or character is present in a string, we can use the keyword in.

```python
txt = "The best things in life are free!"
print("free" in txt)



if "free" in txt:
  print("Yes, 'free' is present.")
```

# Check if NOT
To check if a certain phrase or character is NOT present in a string, we can use the keyword not in.

```python
txt = "The best things in life are free!"
print("expensive" not in txt)

```

Use it in an if statement:
 only if "expensive" is NOT present:

```python
txt = "The best things in life are free!"
if "expensive" not in txt:
  print("No, 'expensive' is NOT present.")
```

# Slicing
You can return a range of characters by using the slice syntax.

Specify the start index and the end index, separated by a colon, to return a part of the string.

Note: The first character has index 0.

```python
b = "Hello, World!"
print(b[2:5])
```

# Slice From the Start
By leaving out the start index, the range will start at the first character:

```python
b = "Hello, World!"
print(b[:5])
```

# Slice To the End
By leaving out the end index, the range will go to the end:

```python
b = "Hello, World!"
#Get the characters from position 2, and all the way to the end:
print(b[2:])
```

# Negative Indexing
Use negative indexes to start the slice from the end of the string:

Get the characters:

From: "o" in "World!" (position -5)

To, but not included: "d" in "World!" (position -2):

```python
text = "Hello, World!"
print(text[-5:-2])
```

Modify Strings
Python has a set of built-in methods that you can use on strings.

```python
text = "Hello, World!"
print(text.upper()) # Convert to upper case
print(text.lower()) # Convert to lower case
print(text.replace("H", "J")) # Replace H with J
print(text.split(",")) # Split the string at the comma
print(text.strip()) # Remove whitespace from the beginning or the end
# print(text.remove("H")) # Remove H from the string
print(text.index("W")) # Get the index of the first occurrence of W
print(text.find("W")) # Find the first occurrence of W
print(text.count("o")) # Count how many times o appears in the string
print(text.startswith("Hello")) # Check if the string starts with Hello
print(text.endswith("World!")) # Check if the string ends with World!
print(text.isalpha()) # Check if all characters in the string are alphabetic
print(text.isdigit()) # Check if all characters in the string are digits
print(text.islower()) # Check if all characters in the string are lowercase
print(text.isupper()) # Check if all characters in the string are uppercase
print(text.isspace()) # Check if all characters in the string are whitespace
print(text.capitalize()) # Capitalize the first character of the string
print(text.title()) # Convert the first character of each word to upper case
print(text.center(50)) # Center the string with a total width of 50
print(text.ljust(50)) # Left justify the string with a total width of 50
print(text.rjust(50)) # Right justify the string with a total width of 50
print(text.zfill(50)) # Pad the string with zeros on the left to a total width of 50
print(text.encode()) # Encode the string to bytes
print(text.format()) # Format the string
print(text.join(["Hello", "World"])) # Join a list of strings with the string as separator
print(text.partition("World")) # Partition the string at the first occurrence of World
print(text.rpartition("World")) # Partition the string at the last occurrence of World
print(text.splitlines()) # Split the string at line breaks
print(text.swapcase()) # Swap the case of the string
print(text.translate(str.maketrans("Helo", "Jalo"))) # Translate characters in the string
print(text.removeprefix("Hello")) # Remove the prefix Hello from the string
print(text.removesuffix("World!")) # Remove the suffix World! from the string



```

```python
a = "Hello"
b = "World"
c = a + b
print(c)
```

# F-Strings
F-String was introduced in Python 3.6, and is now the preferred way of formatting strings.

To specify a string as an f-string, simply put an f in front of the string literal, and add curly brackets {} as placeholders for variables and other operations.

```python
age = 36
txt = f"My name is John, I am {age}"
print(txt)
```

# Placeholders and Modifiers
A placeholder can contain variables, operations, functions, and modifiers to format the value.

```python
price = 59
txt = f"The price is {price} dollars"
print(txt)
```

A placeholder can include a modifier to format the value.

A modifier is included by adding a colon : followed by a legal formatting type, like .2f which means fixed point number with 2 decimals:

```python
price = 59
txt = f"The price is {price:.2f} dollars"
print(txt)

txt = f"The price is {20 * 59} dollars"
print(txt)
```
