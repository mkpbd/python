```python
# You must run this code and use that data set for all the questions
import numpy as np
import pandas as pd

np.random.seed(42)

ids = np.arange(1, 11)
ages = np.random.randint(18, 60, 10)
salaries = np.random.randint(30000, 90000, 10)
departments = np.array(["HR", "IT", "Finance", "IT", "HR", "Sales", "Finance", "IT", "Sales", "HR"])

DF = pd.DataFrame({
"id": ids,
"age": ages,
"salary": salaries,
"dept": departments
})

DF.to_csv("employees.csv", index=False)
print("Sample Data Created and Saved as employees.csv")
```

# 1) NumPy — 5 Questions


Q1. Create NumPy arrays from the ages and salaries data generated above. Print both arrays and display their dtype, ndim, shape, and size.


```python
import numpy as np
ages_array = np.array(ages)
salaries_array = np.array(salaries)

print("Ages Array :", ages_array)
print(f" dtype: {ages_array.dtype}")
print(f"ndim: {ages_array.ndim}")
print(f"shape: {ages_array.shape}")
print(f"size: {ages_array.size}")

print("Salary Array:" , salaries_array)
print(f"dtype: {salaries_array.dtype}")
print(f"ndim: {salaries_array.ndim}")
print(f"shape: {salaries_array.shape}")
print(f"size: {salaries_array.size}")
```

## Q2. Using the salaries array, find and print the highest and lowest salary values. Also, calculate and print the average salary and age using NumPy functions.


```python
import numpy as np
ages_array = np.array(ages)
salaries_array = np.array(salaries)
print(f"Highest Salary: {np.max(salaries_array):,}")
print(f"Lowest Salary: {np.min(salaries_array):,}")
print(f"Average Salary: {np.mean(salaries_array):,.2f}")
print(f"Average Age: {np.mean(ages_array):.2f} years")
```
