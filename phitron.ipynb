Using filter() and lambda, extract even numbers from a given list.
Input Format:
 Integers separated by space.
Output Format:
 List of even integers.
Example:
Input: 5 6 7 8 9 10  
Output: [6, 8, 10]


```python
get_input = input().strip()
convert_number = map(int, get_input.split())
convert_list = list(convert_number)

evenNumber = list( filter(lambda x: x % 2 == 0, convert_list))

print(evenNumber)
```

Write a function safe_divide(a,b) that performs a/b and handles division by zero using try–except.
 Print "Division by zero" if b = 0.
Input Format:
 Two integers a b.
Output Format:
 Result or error message.
Example:
Input: 10 0  
Output: Division by zero


```python
def safe_divide(a, b):
  try:
    result = a/b
    print(result)
  except ZeroDivisionError:
    print("Division by zero")

a, b = map(int, input().split())

safe_divide(a, b)
```

Create a base class Shape with a method area() that returns 0.
 Create subclasses:
Rectangle(w,h) → area = w × h


Circle(r) → area = π × r²


Read shape type and parameters, then print area (rounded to 2 decimals).
Input Format 1:
Rectangle 5 10

Output Format 1:
Area: 50.00


Input Format 2:
Circle 7

Output Format 2:
Area: 153.94


```python
import math
## base Class
class Shape:
  def area(self):
    return 0

## sub class for reactange
class Rectangle(Shape):
  def __init__(self, w, h):
    self.w = w
    self.h = h
  def area(self):
    return self.w * self.h

## Sub Class for Circle
class Circle(Shape):

  def __init__(self, r):
    self.r = r

  def area(self):
    return  math.pi * (self.r ** 2)

## Rectangle object create
rect = Rectangle(5, 10)
area = rect.area()
print(f"Area: {area: .2f}")

## Cricle object Create
circle = Circle(7)

circle_area = circle.area()
print(f"Area: {circle_area: .2f}")
```

Find and print the longest word in a given sentence (if tie, print first one).
Input Format:
 One line string
Output Format:
 Longest word
Example:
Input: Python programming is enjoyable
Output: programming


```python
def longest_word(sentance):
  # sentance convert to list
   word = sentance.split()
   ## use max  function or method  long word  count
   long_word = max(word, key=len)
   return long_word


get_input_string = input().strip()

result = longest_word(get_input_string)

print(result)
```

A number is called strong if the sum of the factorial of its digits equals the number itself.
 Example: 145 = 1! + 4! + 5!
Input Format:
 Single integer
Output Format:
 "Strong Number" or "Not Strong Number"
Example:
Input: 145
Output: Strong Number


```python
import math
def strong_number (number):
    copy_orginal_number = number
    fac_sum = 0;

    ## looping

    while number > 0:
        # get last digit
        digit = number % 10
        # add factorial of digit
        fac_sum += math.factorial(digit)
        number //= 10

    # check the strong number
    if fac_sum == copy_orginal_number:
        print("Strong Number")
    else :
        print("Not Strong Number")

number = int(input().strip())

strong_number(number)

```
