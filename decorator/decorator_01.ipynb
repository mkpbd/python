Decorators let you add extra behavior to a function, without changing the function's code.

A decorator is a function that takes another function as input and returns a new function.

# Basic Decorator
Define the decorator first, then apply it with @decorator_name above the function.

```python
def change_title_case(func):
    def inner_func():
        return func().title()
    return inner_func
    
@change_title_case
def my_title():
    return 'mostofa kamal passa'

print(my_title())
```

```python
def change_case (func):
    def inner_fuc(x):
        return func(x).upper()
    return inner_fuc

@change_case
def my_function(x):
    return x

print(my_function('hello bangaldesh'))
```

```python
## secure the function with  *args  and ** kwargs  arguments 

def  change_case (func):
    def my_inner(*args, **kwargs):
        return func(*args, **kwargs).upper()
    return my_inner

@change_case
def my_function(num):
    return "hello "+ num

print(my_function("Mostofa kamal passa"))
```

```python
## A decorator factory that takes an argument and transforms the casing based on the argument value.

def change_case(n):
    def change_case(func):
        def my_inner_func():
            if(n == 1):
                a = func().lower()
            else:
                a =  func().upper()
            return a 
        return my_inner_func
    return change_case


@change_case(2)
def my_test_function():
    return 'hello bbbb'

print(my_test_function())
```
