## The Difference Between Copy and View

#### The main difference between a copy and a view of an array is that the copy is a new array and the view is just a view of the original array

#### The copy owns the data and any changes made to the copy will not affect original array and any changes made to the original array will not effect the copy 

### the view dose own the data and any change made to the view will affect the original array and any changes made to the original array will affect the view 

`the Copy Should not be affected by the changes made to the original array`

```python
## Copy 
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
arr_copy = arr.copy()
arr_copy[0] = 10
print("Original Array:", arr)
print("Copied Array:", arr_copy)
## View
arr_view = arr.view()
arr_view[1] = 20
print("Original Array after modifying view:", arr)
print("View Array:", arr_view)
```
