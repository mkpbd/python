```python
## Transpose matrix 
import numpy as np

mat = np.array([
    [10, 20, 30, 40, 50],
    [30,40,50, 60,70]
])
## convert to transport matrix

transporse = mat.T

print(transporse)

```

```python
## Array split 

arr = np.random.randint(2,15, size=(15,))
print(arr)
## array split 
arr_split = np.array_split(arr, 4)
print(arr_split )
```

## Arithmetic operator and mathematical functions 

```python
x = np.array([10,20,30])
y = np.array([30, 40, 50])

add = x+y;
print(add)
## minus 
minis = x-y;
print(minis)
## multiplication 
mul = x*y;
print(mul)

```
